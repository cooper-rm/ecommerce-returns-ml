# EDA Profile

ydata-profiling reports (HTML + PDF) for the raw order lines, sampled to 100k rows, written to `eda/eda-raw/`.

**Profiling helper.** ydata-profiling generates distributions, missingness and correlations automatically. Large tables are sampled to 100k rows with a fixed `random_state` — profiling 2.3M rows takes minutes and reveals nothing extra. The Chrome subprocess prints each HTML report to PDF for offline reading.

```python
import sys; sys.path.append("..")
import warnings; warnings.filterwarnings("ignore")
import subprocess
from ydata_profiling import ProfileReport
from src.data import ROOT, load_orders

EDA = ROOT / "eda"
SAMPLE = 100_000
CHROME = "/Applications/Google Chrome.app/Contents/MacOS/Google Chrome"

def report(df, name, outdir):
    outdir.mkdir(parents=True, exist_ok=True)
    if len(df) > SAMPLE:
        df = df.sample(SAMPLE, random_state=0)
    html = outdir / f"{name}.html"
    ProfileReport(df, title=name, explorative=True).to_file(html)
    subprocess.run([CHROME, "--headless=new", "--disable-gpu", "--no-pdf-header-footer",
                    "--virtual-time-budget=60000",
                    f"--print-to-pdf={html.with_suffix('.pdf')}", html.resolve().as_uri()],
                   check=True, capture_output=True)
```

## Raw

**Run it on both raw files.** Outputs go to `eda/eda-raw/` and are gitignored, since they are multi-megabyte HTML. Regenerate locally rather than committing them.

```python
for name in ["orders_train", "orders_class"]:
    report(load_orders(name), name, EDA / "eda-raw")
```
